```python
import matplotlib.pyplot as plt
import math
import matplotlib.patches as patches
import numpy as np
import matplotlib
import matplotlib.animation as anm    #アニメーションを扱うためのモジュール
#matplotlib.use('nbagg')   今のjupyter notebookのバージョンでは使えない
from IPython.display import HTML    #アニメーションを生成するため
```

```python
class World:
    def __init__(self,debug=False):      #デバッグ用のフラグを追加
        self.objects = []   
        self.debug = debug    #アニメーションさせるとエラーメッセージが出ない　debugがTrueになるとアニメーションを切る

    def append(self,obj):   
        self.objects.append(obj)

    def draw(self):
        fig = plt.figure(figsize=(4,4))    
        ax = fig.add_subplot(111)         
        ax.set_aspect('equal')            
        ax.set_xlim(-5,5)                 
        ax.set_ylim(-5,5)                  
        ax.set_xlabel("X",fontsize=10)     
        ax.set_ylabel("Y",fontsize=10)    

        elems = []    #描画する図形のリスト
        
        if self.debug:
            for i in range(1000):self.one_step(i,elems,ax)    #デバッグ時はアニメーションさせない
        else:
            self.ani = anm.FuncAnimation(fig,self.one_step,fargs=(elems,ax),frames=10,interval=1000,repeat=False)
            #anm.FuncAnimationというクラスのオブジェクトを作り変数self.aniと結びつける
            #引数は図のオブジェクト,１ステップ進めるメソッド,one_stepに渡す引数,描画する総ステップ数,ステップの周期,繰り返し描画するかのフラグ
            plt.close()    #plt.show()がなくても静止画が描画されてしまうので静止画を消すために入れる
            return HTML(self.ani.to_jshtml())    #returnがないと描画されない　詳しくはhttps://hooked-on-mas.hatenablog.com/entry/2019/11/16/012422を確認
            #to_jshtmlでアニメーションをHTML表現に変換、HTML()で表示させる

    def one_step(self,i,elems,ax):    #アニメーションを1コマ進めるメソッド  iはステップの番号
        #pass#なにもさせないためのpass
        while elems:elems.pop().remove()    #２重描画を防ぐためにelemsのリストをクリア　　popはリストの一番後ろからオブジェクトを取り出してリストから消す関数
        #removeはelemsに入ってるオブジェクトのメソッドでオブジェクト自身を消す関数　　オブジェクトをリストから消してオブジェクト自身も消す
        elems.append(ax.text(-4.4, 4.5,"t = "+str(i),fontsize=10))    #図の左上に時刻を表示させる　str(i)でステップ番号iを文字に変換
        for obj in self.objects:
            obj.draw(ax,elems)
```

```python
class IdealRobot:
    def __init__(self,pose,color="black"):
        self.pose = pose    
        self.r = 0.2        
        self.color = color  

    def draw(self,ax,elems):    #引数にelemsを追加
        x,y,theta = self.pose    
        xn = x + self.r * math.cos(theta)    
        yn = y + self.r * math.sin(theta)    
        elems += ax.plot([x,xn],[y,yn],color=self.color)    #elemsに追加するため"elems +="を追加　リスト同士の足し算なのはax.plotがリストを返すため
        c = patches.Circle(xy=(x,y),radius=self.r,fill=False,color=self.color)    
        elems.append(ax.add_patch(c))    #elems.append()でロボットをelemsに追加
```

```python
world = World()    #コードにバグがあるときworld=World(debug=True)としてデバッグを行う

robot1 = IdealRobot(np.array([2,3,math.pi/6]).T)    
robot2 = IdealRobot(np.array([-2,-1,math.pi/5*6]).T,"red")    
world.append(robot1)    
world.append(robot2)
world.draw()
```
