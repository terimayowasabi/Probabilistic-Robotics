```python
import matplotlib.pyplot as plt
import math
import matplotlib.patches as patches
import numpy as np
import matplotlib
matpltlib.use('nbagg')    #アニメーションを表示するという宣言 
import matplotlib.animation as anm    #アニメーションを扱うためのモジュール
```

```python
class World:
    def __init__(self,debug=False):      #デバッグ用のフラグを追加
        self.objects = []    #ここにロボットなどのオブジェクトを登録
        self.debug = debug    #アニメーションさせるとエラーメッセージが出ない　debugがTrueになるとアニメーションを切る

    def append(self,obj):    #オブジェクトを登録するための関数
        self.objects.append(obj)

    def draw(self):
        fig = plt.figure(figsize=(8,8))    # 8x8 inchの図を準備   figオブジェクトは画像全体の入れ物
        ax = fig.add_subplot(111)          #サブプロットを準備   (行,列,番目)　図を行、列で分割したn番目に配置
        ax.set_aspect('equal')             #縦横比を座標の値と一致させる
        ax.set_xlim(-5,5)                  #X軸を-5m x 5mの範囲で描画
        ax.set_ylim(-5,5)                  #Y軸を-5m x 5mの範囲で描画
        ax.set_xlabel("X",fontsize=20)     #X軸にラベルを表示
        ax.set_ylabel("Y",fontsize=20)     #Y軸にラベルを表示

        elems = []    #描画する図形のリスト

        if self.debug:
            for i in range(1000):self.one_step(i,elems,ax)    #デバッグ時はアニメーションさせない
        else:
            self.ani = anm.FuncAnimation(fig,self.one_step,fargs=(elems,ax),frames=10,interval=1000,repeat=False)
            #anm.FuncAnimationというクラスのオブジェクトを作り変数self.aniと結びつける
            #引数は図のオブジェクト,１ステップ進めるメソッド,one_stepに渡す引数,描画する総ステップ数,ステップの周期,繰り返し描画するかのフラグ
            plt.show()

        def one_step(self,i,elems,ax):    #アニメーションを1コマ進めるメソッド  iはステップの番号
            pass    #なにもさせないためpass
            

        for obj in self.objects:obj.draw(ax)      #self.objectsにあるオブジェクトのdrawメソッドを次々に呼ぶ出して描画
       #for obj in self.objects:　　↑　for文の中身が１行なら1行にまとめて良い
       #    obj.draw(ax)         　self.objectsの中のオブジェクトにax上に自分を書いてねと命令
        plt.show()
```

```python
class IdealRobot:
    def __init__(self,pose,color="black"):
        self.pose = pose    #引数から姿勢の初期値を設定
        self.r = 0.2        #描画のため固定値　　半径rを0.2に設定
        self.color = color  #引数から描画するときの色を設定　　デフォルト値に黒を指定しているcolor="black"

    def draw(self,ax):
        x,y,theta = self.pose    #姿勢の変数を分解して3つの変数へ
        xn = x + self.r * math.cos(theta)    #ロボットの鼻先のx座標
        yn = y + self.r * math.sin(theta)    #ロボットの鼻先のy座標
        ax.plot([x,xn],[y,yn],color=self.color)    #ロボットの向きを示す線分の描画
        c = patches.Circle(xy=(x,y),radius=self.r,fill=False,color=self.color)    #patches.Circleでロボットの胴体を示す円を作る
        ax.add_patch(c)    #サブプロットに登録　　add_patch()で（）内の図に追加
```

```python
world = World()

robot1 = IdealRobot(np.array([2,3,math.pi/6]).T)    #ロボットのインスタンス生成
robot2 = IdealRobot(np.array([-2,-1,math.pi/5*6]).T,"red")    #色指定込み
world.append(robot1)    #Worldクラスのappendを使用してobjectにrobotを登録
world.append(robot2)
world.draw()
```
