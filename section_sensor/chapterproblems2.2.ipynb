## 章末問題 2.2

```python
import random
import matplotlib.pyplot as plt
```

```python
def dice_noise():       #サイコロを10個(10回)振って合計値に200を足す
    dice = [random.choice([1,2,3,4,5,6]) for i in range(10)]
    return sum(dice)+200
```

```python
dice10_noise = [dice_noise() for i in range(10000)]
plt.hist(dice10_noise,bins= max(dice10_noise)-min(dice10_noise))
plt.show()
```
