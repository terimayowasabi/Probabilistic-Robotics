```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import math 
from scipy.stats import binom

data_2 = pd.read_csv("sensor_data_200.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_28 = pd.read_csv("sensor_data_280.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_3 = pd.read_csv("sensor_data_300.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_4 = pd.read_csv("sensor_data_400.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_5 = pd.read_csv("sensor_data_500.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_6 = pd.read_csv("sensor_data_600.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_7 = pd.read_csv("sensor_data_700.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))
```

```python
def data_mv(data):
    mean = data["lidar"].mean()
    var = data["lidar"].var()
    return mean,var
```

```python
def binomial(data):
    mean = data["lidar"].mean()
    var = data["lidar"].var()

    p = 1-var/mean
    n = round(mean**2/(mean-var))
    return binom(n,p)
```

```python
def xrange(data):
    return range(min(data["lidar"]),max(data["lidar"]))
```

```python
def bayes_estimation(sensor_value):
    new_estimation = []
    for i in (200,280,300,400,500,600,700):
        new_estimation.append(P_f[i].pmf(sensor_value)/7)
        
    return new_estimation/sum(new_estimation)
```
