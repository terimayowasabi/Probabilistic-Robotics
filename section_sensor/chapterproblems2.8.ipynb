```python
import pandas as pd
import matplotlib.pyplot as plt
import math 
from scipy.stats import binom

data_2 = pd.read_csv("sensor_data_200.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_28 = pd.read_csv("sensor_data_280.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_3 = pd.read_csv("sensor_data_300.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_4 = pd.read_csv("sensor_data_400.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

data_5 = pd.read_csv("sensor_data_500.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))
```

```python
def data_mv(data):
    mean = data["lidar"].mean()
    var = data["lidar"].var()
    return mean,var
```

```python
def binomial(data):
    mean = data["lidar"].mean()
    var = data["lidar"].var()

    p = 1-var/mean
    n = round(mean**2/(mean-var))
    return binom(n,p)
```

```python
def xrange(data):
    return range(min(data["lidar"]),max(data["lidar"]))
```

```python
b_2 = binomial(data_2)
x2 = xrange(data_2)
P2 = b_2.pmf(x2)
mean_2,var_2 = data_mv(data_2)

plt.plot(x2,P2)
plt.show()
print(f"""
データファイル:sensor_data_200.txt
平均値:{mean_2}
標本分散:{var_2}
""")
```
