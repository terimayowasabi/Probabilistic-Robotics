```python
def binom_np(mean,var):
    p = 1-var/mean
    n = round(mean**2/(mean-var))
    return binom(n,p)
```

```python
def xrange(data):
    return range(min(data["lidar"]),max(data["lidar"]))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import math 
from scipy.stats import binom

data_2 = pd.read_csv("sensor_data_200.txt", delimiter=' ',
               header=None, names=("deta","time","ir","lidar"))

mean_2 = data_2["lidar"].mean()
var_2 = data_2["lidar"].var()
d_2 = binom_np(mean_2,var_2)

data_2["lidar"].hist(bins=max(data_2["lidar"])-min(data_2["lidar"]),align='left')
plt.show()

print(f"""
データファイル:sensor_data_200.txt
平均値:{mean_2}
分散:{var_2}
""")
```
